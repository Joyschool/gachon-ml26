### [실습 2] Machine Learning: 규칙 대신 데이터를 준다

```python
# practice2_ml.py
# Machine Learning: 규칙 대신 데이터를 준다
import numpy as np
from sklearn.tree import DecisionTreeClassifier, export_text

# 사람이 제공하는 것: (무게 g, 빨간 정도 0~10) 와 정답
X = np.array([[130, 9], [140, 8], [150, 8], [160, 7],    # 사과
              [170, 4], [180, 3], [190, 4], [200, 2]])   # 오렌지
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])   # 0=사과, 1=오렌지

model = DecisionTreeClassifier(max_depth=2)
model.fit(X, y)                        # ← 규칙은 여기서 컴퓨터가 찾는다
print(model.predict([[155, 8]]))       # [0] → 사과
```

###[질문] 정말 컴퓨터가 규칙을 찾았을까?

```python
# 정말 컴퓨터가 규칙을 찾았을까?
#  - 학습된 규칙 열어보기
print(export_text(model, feature_names=["weight", "redness"]))
```

- **Q** :  max_depth=1 로 바꾸면?

```python
model = DecisionTreeClassifier(max_depth=1)
model.fit(X, y)

print(model.predict([[155, 8]]))
print('-'*50)
print(export_text(model, feature_names=["weight", "redness"]))                     # ← 규
```

### **④ Perceptron: 규칙 대신 가중치를 배우자**
- 1958년 Frank Rosenblatt의 **Perceptron**은 입력값을 이용하여 분류를 학습하는 초기 **신경망 모델의 대표적인 사례**
> 공부시간이 많고<br>출석률이 높으면<br>
어떤 결과가 나올 가능성이 높다

- $y=w_1​x_1​+w_2​x_2​+b$
- → $w_1, w_2, b$를데이터에서 찾아내도록 하는데 여기서 **학습한다**는 말의 실체가 나타남.
> 즉, **학습** = 데이터를 보고 모델의 파라미터를 수정하는 과정

### [실습 3] Perceptron으로 '가중치를 학습한다'를 직접 **확인**

```python
# practice3_perceptron.py
# Perceptron으로 '가중치를 학습한다'를 직접 확인
import numpy as np

# AND 데이터
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])

w = np.zeros(2)          # 처음엔 아무것도 모름
b = 0
learning_rate = 0.1

for epoch in range(10):
    for x, target in zip(X, y):
        score = np.dot(x, w) + b     # 두 벡터 x와 w의 내적(dot product)을 계산
        prediction = 1 if score > 0 else 0
        error = target - prediction            # 틀렸는가?
        w = w + learning_rate * error * x      # ★ 학습: 가중치 수정
        b = b + learning_rate * error          # ★ 학습: 편향 수정

        print(f"x={x} w={w} b={b} score={score} prediction={prediction} error={error} w={w} b={b} " )
print('-'* 50)
print("weight =", w)
print("bias   =", b)
# 검증
for x in X:
    print(x, "→", 1 if np.dot(x, w) + b > 0 else 0)
```

- [정리] 1958 퍼셉트론 → 1969 XOR 한계 → 1986 역전파(은닉층) — numpy만으로 재현

```python
# 1958 퍼셉트론 → 1969 XOR 한계 → 1986 역전파(은닉층) — numpy만으로 재현
import numpy as np
rng = np.random.default_rng(0)

X = np.array([[0,0],[0,1],[1,0],[1,1]], dtype=float)
y_and = np.array([0,0,0,1]); y_xor = np.array([0,1,1,0])

# ① Rosenblatt(1958) 퍼셉트론: w·x + b > 0 이면 1, 틀리면 w ← w + η(y-ŷ)x
def perceptron(X, y, epochs=20, lr=0.1):
    w, b = np.zeros(X.shape[1]), 0.0
    for _ in range(epochs):
        for xi, yi in zip(X, y):
            yhat = int(w @ xi + b > 0)
            w += lr * (yi - yhat) * xi;  b += lr * (yi - yhat)
    return np.array([int(w @ xi + b > 0) for xi in X])

print("AND 퍼셉트론:", perceptron(X, y_and), "정답:", y_and)   # 학습 성공
print("XOR 퍼셉트론:", perceptron(X, y_xor), "정답:", y_xor)   # 직선 하나로는 못 가름 (Minsky & Papert 1969)

# ② Rumelhart·Hinton·Williams(1986) 역전파: 은닉층 2개 뉴런 + 시그모이드
sig = lambda z: 1/(1+np.exp(-z))
W1 = rng.normal(0, 1, (2, 2)); b1 = np.zeros(2)
W2 = rng.normal(0, 1, (2, 1)); b2 = np.zeros(1)
Y = y_xor.reshape(-1, 1); lr = 0.5
for step in range(5000):
    h = sig(X @ W1 + b1);  out = sig(h @ W2 + b2)     # 순전파
    d_out = (out - Y) * out * (1 - out)                # 출력층 오차 신호(∂L/∂z)
    d_h = (d_out @ W2.T) * h * (1 - h)                 # 은닉층으로 오차를 '역전파'
    W2 -= lr * h.T @ d_out;  b2 -= lr * d_out.sum(0)   # 경사 하강
    W1 -= lr * X.T @ d_h;    b1 -= lr * d_h.sum(0)
print("XOR 2층 신경망:", (out > 0.5).astype(int).ravel(), "정답:", y_xor)
print("→ '층을 쌓고 오차를 거꾸로 흘리면' 비선형도 배운다 — 딥러닝의 출발점")
```

```python
import re
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import cross_val_score, StratifiedKFold

# ---------------------------------------------------------------
# 데이터: 문자 메시지 20건 (1 = 스팸, 0 = 정상)
# ---------------------------------------------------------------
messages = [
    ("[광고] 무료 쿠폰 지급!! 지금 클릭 http://bit.ly/xx 당첨 확인", 1),
    ("★특별 대출★ 최저금리 1.9% 무서류 즉시 승인!!! 010-1234-5678", 1),
    ("축하합니다! 100만원 상품권 당첨!!! 링크 클릭 http://ab.cd", 1),
    ("(광고) 오늘만 90% 할인!! 무료배송 놓치지 마세요 http://sale.kr", 1),
    ("회원님 계정이 정지됩니다. 즉시 인증 http://bank-secure.xyz 클릭", 1),
    ("무료 체험 이벤트 당첨!! 선착순 100명 지금 신청 010-9999-0000", 1),
    ("[Web발신] 대출 한도 조회 무료!! 3분 승인 http://loan.kr", 1),
    ("★★긴급★★ 택배 미수령 주소 확인 요망 http://cj-delivery.co", 1),
    ("(광고)카지노 첫충 200% 보너스!!! 지금 가입 http://win.bet", 1),
    ("당첨!! 최신 스마트폰 0원 무료 수령 클릭 http://gift.zz", 1),
    ("교수님, 내일 수업 과제 제출 기한이 언제까지인가요?", 0),
    ("오늘 저녁 7시에 학교 앞에서 만나자", 0),
    ("엄마 나 지금 집에 가는 중이야 30분 뒤 도착", 0),
    ("회의 자료 공유드립니다. 검토 후 의견 부탁드려요.", 0),
    ("[Web발신] 주문하신 상품이 오늘 배송 완료되었습니다.", 0),
    ("생일 축하해! 주말에 밥 한번 먹자", 0),
    ("팀 프로젝트 깃허브 링크 https://github.com/team/repo 올렸어", 0),
    ("내일 오전 10시 치과 예약 확인 문자입니다.", 0),
    ("과제 3번 문제 힌트 좀 줄 수 있어? 계속 오류나", 0),
    ("이번 학기 성적 이의신청 기간은 9/15까지입니다.", 0),
]
texts = [m for m, _ in messages]
y = np.array([lab for _, lab in messages])

# ---------------------------------------------------------------
# 실습 1. 사람이 특징을 설계한다 — "스팸은 어떤 모습일까?"를 코드로
# ---------------------------------------------------------------
SPAM_WORDS = ["무료", "당첨", "대출", "광고", "클릭", "할인", "긴급", "즉시"]

def extract_features(text: str) -> dict:
    """사람이 생각한 8가지 특징 (Feature Engineering)"""
    return {
        "length":        len(text),                                   # 글자 수
        "n_exclaim":     text.count("!"),                             # 느낌표 개수
        "n_spam_words":  sum(text.count(w) for w in SPAM_WORDS),      # 스팸 단어 등장 수
        "has_url":       int(bool(re.search(r"https?://", text))),   # URL 포함 여부
        "has_phone":     int(bool(re.search(r"\d{3}-\d{4}-\d{4}", text))),  # 전화번호
        "n_digits":      sum(c.isdigit() for c in text),              # 숫자 개수
        "n_special":     sum(c in "★☆[]()%" for c in text),           # 특수문자 개수
        "has_question":  int("?" in text),                            # 물음표 (대화체)
    }

X_hand = pd.DataFrame([extract_features(t) for t in texts])
X_hand["label"] = y
print("=== 실습 1. 사람이 설계한 특징 (앞 6행) ===")
print(X_hand.head(6).to_string())
print("\n스팸/정상 평균 비교:")
print(X_hand.groupby("label").mean().round(2).to_string())
```

- **2.설계한 특징으로 ML 학습**
    - → 컴퓨터가 찾은 규칙 확인
    - StratifiedKFold()는 기존 k-fold 분할 방식에서 특정 클래스(정답) 데이터가 한쪽으로 쏠리는 것을 방지하기 위해 보완된 교차 검증 분할기
    - 일반적인 K-Fold는 단순히 데이터를 K개로 나누지만, StratifiedKFold는 전체 데이터의 정답 클래스 비율을 고려하기 때문에 보통 분류 문제일 경우에 널리 사용됨.

```python
# ---------------------------------------------------------------
# 실습 2. 설계한 특징으로 학습 → 컴퓨터가 찾은 규칙 보기
# ---------------------------------------------------------------
# 의사결정트리 = 해석(규칙 보기),
X_h = X_hand.drop(columns=["label"])
tree = DecisionTreeClassifier(max_depth=2, random_state=0).fit(X_h, y)
print("\n=== 실습 2. Decision Tree 가 특징에서 찾은 규칙 ===")
print(export_text(tree, feature_names=list(X_h.columns)))

# CV 정확도 = 검증(얼마나 잘 맞나)
# 로지스틱 회귀(Logistic Regression) 모델을 교차 검증(Cross Validation) 방식으로 평가하고, 그 평균 정확도를 구하여 출력
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
acc_hand = cross_val_score(LogisticRegression(max_iter=1000), X_h, y, cv=cv).mean()
print(f"수작업 특징 8개 + LogisticRegression  CV 정확도 = {acc_hand:.2f}")

# 새 메시지에 적용하여 스팸여부 예측
new = ["무료 영화 예매권 당첨!! 지금 클릭 http://mv.kr", "오늘 회의 3시로 변경됐어요"]
print("새 메시지 예측:", tree.predict(pd.DataFrame([extract_features(t) for t in new])), "(1=스팸)")
```

- **3.Bag-of-Words / TF-IDF** — '단어 하나 = 특징 하나'로 특징 설계를 반자동화
    - **TF-IDF**(Term Frequency–Inverse Document Frequency)는 "어떤 단어가 이 문서에서는 자주 나오지만, 다른 문서에는 잘 안 나오는가"를 점수로 매기는 방법
    - 단순 단어 빈도(BoW)는 "은/는/이/가"처럼 어디에나 있는 단어에 높은 점수를 주는 문제가 있는데, TF-IDF는 이런 흔한 단어를 자동으로 점수를 깎아 줌

```python
# ---------------------------------------------------------------
# 실습 3. Bag-of-Words / TF-IDF — 특징 설계를 반자동화
#   사람이 '어떤 단어를 볼지' 고르는 대신, 등장한 모든 단어를 특징으로
# ---------------------------------------------------------------
vec = TfidfVectorizer(token_pattern=r"(?u)\b\w+\b")   # 한 글자 토큰도 포함
X_tfidf = vec.fit_transform(texts)
print(f"\n=== 실습 3. TF-IDF: {X_tfidf.shape[0]}개 문장 × {X_tfidf.shape[1]}개 단어 특징 ===")
acc_tfidf = cross_val_score(LogisticRegression(max_iter=1000), X_tfidf, y, cv=cv).mean()
print(f"TF-IDF 특징 {X_tfidf.shape[1]}개 + LogisticRegression  CV 정확도 = {acc_tfidf:.2f}")

# 어떤 단어가 스팸 쪽으로 무게가 실렸나 (사람이 고른 SPAM_WORDS 와 비교)
# clf.coef_[0]는 로지스틱 회귀가 학습한 가중치 벡터(w), coef: coefficient(계수)
clf = LogisticRegression(max_iter=1000).fit(X_tfidf, y)
coef = pd.Series(clf.coef_[0], index=vec.get_feature_names_out()).sort_values()
print("스팸 방향 상위 단어:", list(coef.tail(8).index[::-1]))
print("정상 방향 상위 단어:", list(coef.head(8).index))
```

- **3.Color histogram · Texture(LBP, GLCM)**  — "색 분포와 결(무늬)을 보자"

```python
# ---------------------------------------------------------------
# 실습 3. Color histogram · Texture (LBP: Ojala 2002, GLCM: Haralick 1973)
# ---------------------------------------------------------------
def color_hist(rgb, bins=8):
    """채널별 히스토그램을 이어붙인 3×bins 차원 색 특징"""
    return np.concatenate([np.histogram(rgb[..., c], bins=bins, range=(0, 255))[0] for c in range(3)]).astype(float)

def lbp_hist(gray_u8, P=8, R=1):
    """Local Binary Pattern 히스토그램 (P+2 차원, uniform)"""
    lbp = local_binary_pattern(gray_u8, P, R, method="uniform")
    h, _ = np.histogram(lbp, bins=P + 2, range=(0, P + 2))
    return h / h.sum()

def glcm_feats(gray_u8):
    """GLCM 통계 4종: contrast, homogeneity, energy, correlation"""
    g = graycomatrix(gray_u8, distances=[1], angles=[0, np.pi / 4, np.pi / 2, 3 * np.pi / 4],
                     levels=256, symmetric=True, normed=True)
    return np.array([graycoprops(g, p).mean() for p in ["contrast", "homogeneity", "energy", "correlation"]])

ch = color_hist(img_rgb)
lbp_img = local_binary_pattern(util.img_as_ubyte(img), 8, 1, method="uniform")
print(f"[실습 3] Color 특징 = {len(ch)}차원, LBP 히스토그램 = {len(lbp_hist(util.img_as_ubyte(img)))}차원, GLCM = 4차원")

# ---- 그림 1: 특징 6종 한눈에 ----
fig, ax = plt.subplots(2, 3, figsize=(12, 8))
ax[0, 0].imshow(img_rgb); ax[0, 0].set_title("원본 (Raw Image)")
ax[0, 1].imshow(edges, cmap="gray"); ax[0, 1].set_title(f"Edge — Canny ({edges.sum():,} px)")
ax[0, 2].imshow(img, cmap="gray"); ax[0, 2].plot(corners[:, 1], corners[:, 0], "r.", ms=5); ax[0, 2].set_title(f"Corner — Harris ({len(corners)}개)")
ax[1, 0].imshow(exposure.rescale_intensity(hog_img, in_range=(0, 10)), cmap="gray"); ax[1, 0].set_title(f"HOG ({len(hog_vec)}차원)")
ax[1, 1].imshow(img, cmap="gray"); kp = sift.keypoints[:300]; ax[1, 1].scatter(kp[:, 1], kp[:, 0], s=8, c="cyan"); ax[1, 1].set_title(f"SIFT 키포인트 ({len(sift.keypoints)}개 중 300)")
ax[1, 2].imshow(lbp_img, cmap="gray"); ax[1, 2].set_title("Texture — LBP (uniform, P=8)")
for a in ax.ravel(): a.axis("off")
plt.tight_layout(); plt.show()

# ---- 그림 2: 색 히스토그램 ----
fig, ax = plt.subplots(1, 2, figsize=(9, 3.4))
ax[0].imshow(img_rgb); ax[0].axis("off"); ax[0].set_title("원본")
for c, name in enumerate(["R", "G", "B"]):
    ax[1].plot(ch[c * 8:(c + 1) * 8], marker="o", label=name, color={"R": "#F43F5E", "G": "#10B981", "B": "#3B82F6"}[name])
ax[1].set_title("Color histogram (8 bins × 3채널 = 24차원)"); ax[1].set_xlabel("밝기 구간"); ax[1].set_ylabel("픽셀 수"); ax[1].legend()
plt.tight_layout(); plt.show()
```

### [실습] CNN — 손글씨 digits를 합성곱 신경망으로 분류하기 - 특징까지 학습하기
- 참고 사이트: https://transcranial.github.io/keras-js/#/mnist-cnn

```python
import torch, torch.nn as nn                     # 실습 4의 digits·shift·split 재사용
to_t = lambda I: torch.tensor(I / 16., dtype=torch.float32)[:, None]   # (N,1,8,8)
Xtr, Xte, Xte_shift = to_t(Itr), to_t(Ite), to_t(Ite_shift)

cnn = nn.Sequential(                              # ★ hog16() 같은 역할을 하는 레이어 구현
    nn.Conv2d(1, 16, 3, padding=1), nn.ReLU(),    # 16개 3×3 필터 = '학습되는 Edge 검출기'
    nn.MaxPool2d(2),                              # 8×8 → 4×4 : 위치 흔들림 흡수
    nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(),   # 필터의 조합 = '학습되는 HOG'
    nn.AdaptiveAvgPool2d(1), nn.Flatten(),        # 32차원 특징 벡터 (HOG도 32차원)
    nn.Linear(32, 10))                            # 마지막 한 줄만 분류기

opt = torch.optim.Adam(cnn.parameters(), lr=1e-2)
for epoch in range(30):
    for i in range(0, len(Xtr), 64):
        loss = nn.functional.cross_entropy(cnn(Xtr[i:i+64]), torch.tensor(ytr[i:i+64]))
        opt.zero_grad(); loss.backward(); opt.step()   # 필터 값 = 파라미터 → 조금씩 변경

acc = lambda X, y: (cnn(X).argmax(1).numpy() == y).mean()
print("CNN", acc(Xte, yte), acc(Xte_shift, yte))

# 도전: Xtr에도 shift를 넣어(Data Augmentation) 다시 학습 → 회복되는가?
```
